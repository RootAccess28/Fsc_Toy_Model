#### D.3. Python Data Structures

**Sets**

* A **set** object is an unordered collection of distinct objects.

**Key Characteristics:**
* **No Duplicates:** Sets automatically remove any duplicate values.
* **Unordered:** Unlike lists or modern dictionaries, sets do not maintain insertion order and do not support indexing or slicing.
* **Requirement for Elements:** Elements within a set must be immutable (e.g., strings, numbers, or tuples). However, the set itself is mutable.

**Common usage:**
* Removing duplicates from a sequence
* Mathematical operations such as intersection, union, difference, and symmetric difference.

```python
s = set([1, 5.0, 4.0, 1, 2, 3, 4, 5.0, 'string', True, False, 0])
s
```

* <span style="color:blue;"> The set only keeps one, because 1 and True as well as False and 0 are seen as duplicates
* <span style="color:blue;"> The first added version of the duplicate is typically kept. Therefore, False is kept instead of 0

#### D.3. Python Data Structures

**Dictionary**

A **dictionary** is a built-in data structure used as an **associative array**.

**Key Characteristics:**
*   **Indexing by Key:** Unlike sequences (lists/tuples) that use integer indices, dictionaries use **unique keys**.
*   **Key Requirements:** Keys must be **immutable** objects (e.g., strings, numbers, or tuples).
*   **Values:** A dictionary can contain any Python object as its values.
*   **Ordered:** In modern Python (3.7+), dictionaries maintain the **order** in which items are inserted.
*   **Mutable:** Dictionaries themselves are **mutable**, meaning you can change their content after creation.

#### Using Built-in module: itertools and writing a more abstract loop!

Imagine you have a list of workers and you want to assign them shifts in order.

```python
# Less abstract more convoluted approach by using index

shifts = ["Morning", "Afternoon", "Evening"]
workers = ["Alice", "Bob", "Charlie", "David", "Eve"]

for i in range(len(workers)):
    shift = shifts[i % len(shifts)] # How we are assigning shift to worker is messy
    print(f"{workers[i]}: {shift}")
```

```python
import itertools

# More abstract and advisable use for a loop
# Using iterators such as itertools.cycle(), zip()

shifts = itertools.cycle(["Morning", "Afternoon", "Evening"]) # repeat the sequence indefinitely
workers = ["Alice", "Bob", "Charlie", "David", "Eve"]

# zip() takes one from workers and one from the infinite shifts.
# zip() helps create the iterable we need!
for worker, shift in zip(workers, shifts):
    print(f"{worker}: {shift}")
```

```python
#Define a function
def fibonacci(n):
    if n <= 0:
        return []
    if n==1:
        return [0]
    fib = [0,1]

    for i in range(n-2):
        new_value = fib[-2] + fib[-1]
        fib.append(new_value)
    return fib
    
```

### While loop

What if we want to execute a block of code multiple times but we don't know exactly how many times?

This is a situation when a **while** loop is useful.

```python
#Example:
n = 5
while n > 0:
    print(n)
    n = n - 1
```

<span style="color:blue;">The main points to observe are:

* `while` keyword
* a *logical expression* followed by a colon `:`
* loop executes its code block if the logical expression evaluates to `True`
* update the variable in the logical expression each time through the loop
* **BEWARE!** If the logical expression always evaluates to `True`, then you get an **infinite loop!**

<span style="color:blue;"> We prefer `for` loops over `while` loops because of the last point.

```python
listnum = [x for x in range(7)]
listnum
```

### Example

Imagine you have a list of temperatures in Fahrenheit from a sensor. Some are faulty (extreme outliers like -999°F), and you need them in Celsius for your report.

```python

raw_temps = [72.0, 68.5, -999.0, 75.2, 212.0] # -999.0 is a sensor error

# Initialize an empty list
cleaned_celsius = []

for f in raw_temps:
    if f > -459.67:  # Filtering out the sensor error that falls below absolute zero
        # conversion from Fahrenheit to Celsius
        c = (f - 32) * 5/9
        cleaned_celsius.append(round(c, 2))

print(cleaned_celsius)
```

```python
def f_to_c(f_temp):
    """Function to convert Fahrenheit to Celsius"""
    return (f_temp - 32) * 5/9

def is_reliable(reading):
    """Function to check sensor error"""
    return reading > -459.67

# The scientific pipeline
valid_celsius = [round(f_to_c(t), 2) for t in raw_temps if is_reliable(t)]

valid_celsius
```
